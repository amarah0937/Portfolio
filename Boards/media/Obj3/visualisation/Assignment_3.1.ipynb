```python
#import libraries
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\theba\Downloads\Titanic.csv")
df.head()

class_survival = df.groupby("class")["survived"].mean()
plt.bar(class_survival.index, class_survival.values, color=["#201C52", "#00DBA8", "#02E6ED"])
plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.show()
```

```python
#Were the passengers in higher classes more likely to survive? 
#Yes, passengers in first class survived more often than passenegers in second and third class.
```

```python
#import libraries
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\theba\Downloads\Titanic.csv")
df.head()

df["age_group"] = pd.cut(df["age"], bins=[0, 12, 18, 40, 60, 100],
                         labels=["Child", "Teen", "Adult", "Middle Aged", "Senior"])
# calculate survival rate by age group and gender
age_gender_survival = df.groupby(["age_group", "sex"])["survived"].mean().unstack()

#plot
age_gender_survival.plot(kind="bar", figsize=(8, 5), color=["#201C52", "#00BDA8"])
plt.title("Survival Rate by Age Group and Gender")
plt.xlabel("Age Group")
plt.ylabel("Survival Rate")
plt.show()
```

```python
#Were certain age groups or genders more likely to survive?
#Yes, seniors were more likely to survive and women were more likely to survive
```
